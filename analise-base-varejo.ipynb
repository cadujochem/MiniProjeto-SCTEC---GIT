## Limpeza dos Dados e Transformação de dados

```python
#Conforme verificado nos comandos acima, que o DataFrame possui 4 colunas sem nenhum dado, estas serao removidas.

#Exclui as colunas com todas as linhas vazias
df = df.dropna(axis=1, how="all")

#Conferir exclusão das colunas
df
```

```python
#Alteração das Descrições das colunas - Para melhorar legibilidade

df = df.rename(columns= {
    "DATA": "Data",
    "CO_ID":"Compra_ID",
    "CL_ID":"Cliente_ID",
    "CL_GENERO":"Cliente_Genero",
    "CL_EC":"Cliente_Estado_Civil",
    "CL_FHL":"Cliente_Filhos",
    "CL_SEG":"Cliente_Classe",
    "PR_ID":"Produto_ID",
    "PR_CAT":"Produto_Categoria",
    "PR_NOME":"Produto_Nome"

})

print(df)
```

```python
#Conforme Verificado no comando df.dtypes que a data está como tipo STR devemos altera-la para o tipo Datetime do pandas
df["Data"] = pd.to_datetime(df["Data"], format="%d/%m/%Y")
```

```python
#Conferindo se alterou para Datetime
print(df["Data"].dtype)
```

```python
#Codigo para verificar todas as colunas que possui tipo de dados como STRING para 
# que elimine espaços em brancos, e coloque o formato para apenas primeira letra maiuscula

colunas_texto = [
    "Cliente_Genero",
    "Cliente_Classe",
    "Produto_Categoria",
    "Produto_Nome"
]

for coluna in colunas_texto:
    df[coluna] = df[coluna].str.strip().str.title()
```

```python
#Converter para o tipo Numeric do pandas as colunas com dados em formato Inteiro


colunas_inteiras = [
    "Compra_ID",
    "Cliente_ID",
    "Cliente_Filhos",
    "Produto_ID"
]

for coluna in colunas_inteiras:
    df[coluna] = pd.to_numeric(
        df[coluna],
        errors="coerce"
    )
```

```python
#Troca os Dados da Coluna Cliente_Genero de "F" e "M" para "Feminino" e "Masculino"

df["Cliente_Genero"] = df["Cliente_Genero"].replace({
    "M": "Masculino",
    "F": "Feminino"
})
```

```python
#Conferindo alteração do dado para não abreviado
df["Cliente_Genero"].value_counts()
```

```python
df["Cliente_Estado_Civil"] = df["Cliente_Estado_Civil"].replace({
    1: "Casado/União estável",
    2: "Divorciado",
    3: "Separado",
    4: "Solteiro",
    5: "Viúvo"
}).astype("category")
```

```python
#Conferindo a alteração para categorias das colunas selecionadas na celula anterior
df.dtypes
```

```python
# Checando se com a conversão dos dados para numeric, nenhum dado ficou como nulo devido um preenchimento errado.
# 

df.isna().sum()
```

```python
#Separando Ano, Mes e Dia do formato DateTime em Colunas para coletar informaçãos possíveis fornecer futura analise.
df["Ano"] = df["Data"].dt.year
df["Mes"] = df["Data"].dt.month
df["Dia"] = df["Data"].dt.day
```

```python
#Visualizar novas colunas Ano, Mes e Dia criadas 

df
```

#### Tratamento de Dados Invalidos

```python
#Após verificar que existiam preenchimentos com dados e valores invalidos nas Colunas com formato de Dados String. 
# Identifiquei os possíveis preenchimentos errados e inclui uma variável agrupando todos eles.

valores_invalidos = ['#N/D', '#n/d', 'N/A', 'n/a', 'null', 'NULL', '', ' ']


#Codigo que percorre o DataFrame e apresenta as linhas com valores cadatrados como invalidos encontrado em todas as colunas ").

df_invalidos = df[df.isin(valores_invalidos).any(axis=1)]

#Apresenta as linhas com dados invalidos.
print(df_invalidos)
```

```python
#Verificado que as colunas "Produto_Categoria" e "Produto_Nome" possuiam preenchimentos com os dados invalidos cadastrados na variavel "valores_invalidos"

#Tramento dos invalidos na coluna "PR_CAT" para um Descritivo que possa ser computado na analise.
df["Produto_Categoria"] = df["Produto_Categoria"].apply(
    lambda x: 'Sem Categoria' if pd.isna(x) or str(x) in valores_invalidos else x
)

#Confirmar tratamento dos dados invalidos
categoria_desconhecida = df[df["Produto_Categoria"] == "Sem Categoria"]
print(categoria_desconhecida)

```

```python
#Tramento dos invalidos na coluna "Produto_Nome" para um descritivo que possa ser computado na analise.
df["Produto_Nome"] = df["Produto_Nome"].apply(
    lambda x: 'Produto não cadastrado' if pd.isna(x) or str(x) in valores_invalidos else x
)

#Confirmar tratamento dos dados invalidos
produto_nao_cadastrado = df[df["Produto_Nome"] == "Produto não cadastrado"]
print(produto_nao_cadastrado)
```

```python
#Conferindo que o Produto ID 107 - Identificado anteriormente como sem preenchimento estão preenchidos com os dados tratados
df[df["Produto_ID"] == 107]
```

```python
#Confirmando que todos os valores nulos foram alterados para Não informado ou Produto não cadastrado. Resultado deve apresentar todas as colunas com 0
df.isna().sum()
```

```python
### Transformandos as colunas em categorias.
df["Cliente_Genero"] = df["Cliente_Genero"].astype("category")
df["Cliente_Estado_Civil"] = df["Cliente_Estado_Civil"].astype("category")
df["Cliente_Classe"] = df["Cliente_Classe"].astype("category")
df["Produto_Categoria"] = df["Produto_Categoria"].astype("category")
```

#### Dados Duplicados (Optado por ignorar)

```python
#Comando para verificar em todo o DataFrame se possui preencimento com linhas duplicadas, 
# Como se trata de uma Tabela onde não contabiliza quantidade de itens, não significa necessáriamente que 
# os dados duplicados estão errados,  o cliente pode ter comprado duas vezes o mesmo produto por exemplo.

df.duplicated().sum()
```

```python
#Comando para apresentar as linhas duplicadas.
#Podemos veriricar ao executar que na coluna de Nome dos produtos são produtos que podem haver mais do que uma compra, justificando a não exclusão.
df[df.duplicated()]
```

```python
### Verificando Possíveis OutLiers nos dados de filhos. Para Garantir que alguem não digitou um numero erroneamente e quebre o contexto da analise.

Q1 = df["Cliente_Filhos"].quantile(0.25)
Q3 = df["Cliente_Filhos"].quantile(0.75)

IQR = Q3 - Q1

limite_inferior = 0
limite_superior = Q3 + 1.5 * IQR


print(limite_inferior)
print(limite_superior)


outliers = df[
    (df["Cliente_Filhos"] < limite_inferior) |
    (df["Cliente_Filhos"] > limite_superior) 
]

outliers
```

## Analises e Estatisticas


```python
#Quantidades de Compras cadastradas - Incluindo todos os Itens no mesmo Id de Compra

quantidade_compras = df["Compra_ID"].nunique()

print(f"Quantidade de compras únicas: {quantidade_compras}")
```

```python
filhos = df["Cliente_Filhos"]

print("\n===== ESTATÍSTICA DESCRITIVA =====")

print(f"Contagem: {filhos.count()}")
print(f"Média: {filhos.mean():.2f}")
print(f"Mediana: {filhos.median():.2f}")
print(f"Desvio padrão: {filhos.std():.2f}")
print(f"Máximo: {filhos.max()}")
print(f"Mínimo: {filhos.min()}")

```

```python
print(df.groupby("Cliente_Genero").size())


```

```python
## Percentual de Compras Masculino e Feminino

percentual = df["Cliente_Genero"].value_counts(normalize=True) * 100

print(f"Feminino: {percentual['Feminino']:.2f}%")
print(f"Masculino: {percentual['Masculino']:.2f}%")
```

```python
# Quantidade de vendas por Categoria;
vendas_categoria = (
    df.groupby("Produto_Categoria")
      .size()
      .sort_values(ascending=False)
)

print("\n===== REGISTROS POR CATEGORIA =====")
print(vendas_categoria)
```

```python
vendas_produto = (df.groupby("Produto_Nome")
      .size()
      .sort_values(ascending=False)
)

print("\n===== REGISTROS POR CATEGORIA =====")
print(vendas_produto)
```

```python
# Agrupamento por Genero e Produto - Não percebeu-se tendencia de um genero ter preferencia por alguma categoria que se destaque.
genero_categoria = (df.groupby([
        "Cliente_Genero",
        "Produto_Categoria"
    ])
    .size()
    .sort_values(ascending=False)
)

print("\n===== GÊNERO X CATEGORIA =====")
print(genero_categoria)
```
